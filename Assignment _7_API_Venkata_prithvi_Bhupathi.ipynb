```python
# : What is a RESTful API?
'''
A RESTful API is a way for two systems to talk to each other over the web using simple rules.

-It uses standard HTTP methods like GET, POST, PUT, and DELETE.

-Each piece of data is treated as a “resource,” and every resource has a unique URL.

-REST promotes a stateless design, meaning the server does not store session data; every request carries 
full information.

-Responses are usually in JSON format, making them lightweight and easy to read.

-RESTful APIs are popular because they are easy to build, scale, and integrate with mobile or 
web applications.

'''
```

```python
#2. What is Flask, and why is it popular for building APIs?
'''
Flask is a lightweight Python web framework used for creating APIs and web applications.

-It is known as a “micro-framework” because it only includes essential features, keeping things simple.

-Developers can add extra features when needed using extensions.

-It’s easy to learn, has clean syntax, and works well for small to medium-sized projects.

-Flask provides flexibility—developers can structure the project in any way they like.

-It integrates well with tools like Flask-SQLAlchemy and Flask-RESTful.

-Its simplicity, strong community support, and minimal setup make it a popular choice for API development
'''
```

```python
#3.What are HTTP methods used in RESTful APIs?
'''
-GET – Used to fetch or read data from the server.

-POST – Used to create new data on the server.

-PUT – Used to completely update an existing resource.

-PATCH – Used to partially update a resource.

-DELETE – Used to remove a resource.

-HEAD – Same as GET but returns only headers, not the body.

=OPTIONS – Returns supported methods for a resource.

These methods help define clear rules for API interactions, making communication predictable and
consistent.

'''
```

```python
#4.What is the purpose of the @app.route() decorator in Flask?
'''
-It maps a URL path to a specific function in your Flask application.

-When a user visits that URL, the decorated function gets executed.

-You can specify which HTTP methods the route accepts (GET, POST, etc.).

-It helps organize endpoints neatly within a single file or across modules.

-It acts as the main entry point for defining API routes in Flask.

-Without @app.route(), Flask wouldn’t know which function handles which URL.
'''
```

```python
#5 What is the role of Flask-SQLAlchemy?
'''
Flask-SQLAlchemy is an extension that makes working with databases easier in Flask.

-It provides an ORM (Object Relational Mapper) so you can work with tables using Python classes.

-It helps handle queries, relationships, and table creation without writing raw SQL.

-It integrates smoothly with popular databases like SQLite, MySQL, and PostgreSQL.

-It reduces boilerplate code and improves readability.

-It also handles session management and ensures safe database operations
'''
```

```python
#7.How do you return JSON responses in Flask?
'''
-Import jsonify from Flask.

-Create a dictionary representing your response data.

-Return the JSON using jsonify().

Example:

from flask import jsonify

@app.route("/data")
def data():
    return jsonify({"message": "Success", "code": 200})


Flask automatically sets the correct content-type header: application/json.

JSON responses are commonly used in REST APIs for easy client-side consumption.
'''
```

```python
#8. How do you handle POST requests in Flask?
'''
-Allow POST method in the route using methods=["POST"].

-Import and use request to access JSON or form data.

Example:

from flask import request

@app.route("/submit", methods=["POST"])
def submit():
    data = request.json
    return {"received": data}, 201


-You can read form data using request.form, JSON using request.json, and files using request.files.

-POST is mostly used for creating new records.
'''
```

```python
#9.How do you handle errors in Flask (e.g., 404)?
'''
-Use custom error handlers using @app.errorhandler().

Example:

@app.errorhandler(404)
def not_found(e):
    return {"error": "Page not found"}, 404


-You can define handlers for 400, 403, 500, etc.

-Flask automatically uses these custom handlers whenever the specified error occurs.

-This helps maintain cleaner and user-friendly error messages.
'''
```

```python
#10.How do you structure a Flask app using Blueprints?
'''
-Blueprints help divide a large Flask application into smaller, modular components.

-Create a folder (e.g., users) and inside it create a routes.py file.

-Define a Blueprint:

from flask import Blueprint
users = Blueprint("users", __name__)

@users.route("/profile")
def profile():
    return "User profile"

In the main app file, register the Blueprint:

from users.routes import users
app.register_blueprint(users, url_prefix="/users")

-Each blueprint has its own routes and views, defined in separate files.
-In the main app factory, you register each blueprint with app.register_blueprint().
This keeps routes organized and the project easier to maintain.

'''
```
