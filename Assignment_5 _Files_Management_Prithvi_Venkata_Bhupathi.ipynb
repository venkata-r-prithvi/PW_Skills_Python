```python
#11.What is memory management in Python?
'''
Python manages memory "automatically", so programmers do not need to manually allocate 
or deallocate memory.

Memory is divided into stack (for local variables and function calls) and heap 
(for objects and data structures).

Python uses reference counting: each object keeps track of how many variables refer to it; 
when the count drops to zero, the memory is freed.

Python also uses a garbage collector to handle circular references that reference counting alone 
cannot clean up.

'''
```

```python
#15.How does Python's garbage collection system work
'''
Python uses automatic garbage collection to reclaim memory from objects that are no longer needed 
or referenced.

- The main mechanism is "reference counting": each object keeps track of how many variables refer to it,
when the "count drops to zero", the object is immediately deleted.

-For objects involved in "circular references" (where objects refer to each other), Python uses a 
generational garbage collector to detect and clean up these cycles.

-Objects are divided into "three generations" based on "how long they have existed"; younger 
objects are checked more frequently because they are more likely to become unused.

-The "garbage collector" runs automatically when the "number of object allocations" exceeds a 
certain threshold, but it can also be triggered manually using the gc.collect() function.

-This system helps prevent memory leaks and ensures efficient use of memory, making 
Python programs more reliable.

'''
```

```python
#18.What is the difference between os.fork() and multiprocessing in Python?
'''
os.fork() is a low-level system call available only on Unix-like systems; it creates a child process by 
duplicating the parent process, including all its memory and state.

multiprocessing is a high-level Python module that works on both Unix and Windows; it creates new
processes using different start methods (fork, spawn, or forkserver) depending on the platform.

With os.fork(), the child process inherits everything from the parent, which can be memory-intensive and 
risky if the parent has open files or locks.

multiprocessing provides better portability and safety, especially on Windows, where os.fork() is not
available.

multiprocessing offers easier communication and synchronization between processes using queues, pipes, 
and shared memory.

os.fork() gives more direct control but requires careful handling of resources and is not suitable for
all situations, especially with threads.

'''
```

```python
#18 How would you implement a basic logging setup that logs to a file with rotation after 1MB?
from logging.handlers import RotatingFileHandler

# Set up logging to rotate after 1MB (keep 3 backups)
handler = RotatingFileHandler("app.log", maxBytes=1024*1024, backupCount=3)
handler.setFormatter(logging.Formatter('%(asctime)s - %(levelname)s - %(message)s'))

logger = logging.getLogger()
logger.setLevel(logging.INFO)
logger.addHandler(handler)

# Example log messages
logger.info("Test info")
logger.error("Test error")
```

```python
#21 Write a Python program that reads a file and prints the number of occurrences of a specific word?
count = 0
with open("file_w.txt", "r") as f:
    content = f.read()
    words = content.split() # we need to split the word to fine the specific word.

for w in words:
    if w == "program":
        count += 1

if count > 0:
    print(f"The word '{w}' is repeated {count} time(s)")
else:
    print(f"The word '{w}' is not found")
```
